```python
from pathlib import Path
import decoder as D
# Loop through all .exp files in Season_1 folder
season_folder = Path('Season_1')
exp_files = sorted(season_folder.glob('*.exp'))

for exp_file in exp_files:
    output_name = exp_file.stem + '.json'  # Get filename without extension
    output_path = f'Season1_JSON/{output_name}'
    
    print(f'Processing: {exp_file} -> {output_path}')
    D.main([str(exp_file), '--format', 'json', '-o', output_path])
```
