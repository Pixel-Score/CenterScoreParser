```python
from pathlib import Path
import decoder as D

season_folder = Path('Episodes')
out_folder = Path('JSON')
out_folder.mkdir(parents=True, exist_ok=True)   # <-- add this

exp_files = sorted(season_folder.glob('*.exp'))
for exp_file in exp_files:
    output_path = out_folder / (exp_file.stem + '.json')
    print(f'Processing: {exp_file} -> {output_path}')
    D.main([str(exp_file), '--format', 'json', '-o', str(output_path)])
```
